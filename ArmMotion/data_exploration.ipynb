# Data Exploration

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
pd = pd.read_csv('arm_angles.csv')

plt.figure(figsize=(12, 6))
plt.plot(pd['time'], pd['shoulder_flexion'], label='Shoulder Flexion')
plt.plot(pd['time'], pd['shoulder_abduction'], label='Shoulder Abduction')
plt.plot(pd['time'], pd['elbow_flexion'], label='Elbow Flexion')
plt.title('Arm Angles Over Time')
plt.xlabel('Time (s)')
plt.ylabel('Angle (degrees)')
plt.legend()
plt.grid()
plt.show()
```
